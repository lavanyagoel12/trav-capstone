```python
import pandas as pd
from sklearn.neighbors import NearestNeighbors
import random
import json
import pickle
```

```python
def flatten_json(json_file_path):
    with open(json_file_path, 'r') as f:
        data = json.load(f)

    df_formatted = pd.json_normalize(data)
    return df_formatted

file_path = 'C:/Capstone/trav-capstone/server/python/spotify_formatted.json'
df_read = flatten_json(file_path)
display(df_read)
```

```python
#df2 = df_read.select_dtypes(exclude=['object'])
dfSmall = df_read.groupby('country').apply(lambda x: x.sample(frac=0.28)).reset_index(drop=True)
display(dfSmall)
```

```python
#df_read = pd.read_json('spotify_formatted.json')
# List of columns to keep
'''
columns_to_keep = [
    "artist_genre","record_name","danceability", "energy", "key", "mode", "loudness", "speechiness",
    "acousticness", "instrumentalness", "liveness", "valence", "tempo",
    "duration", "country", "region", "language", "price", "condition", "popularity"
]
'''
columns_to_keep = [
    "danceability", "energy", "key", "mode", "loudness", "speechiness",
    "acousticness", "instrumentalness", "liveness", "valence", "tempo",
    "duration", "price", "condition", "popularity"
]
# Update the DataFrame to keep only the specified columns
df_filtered = dfSmall[columns_to_keep]
display(df_filtered)
```

```python
# Convert categorical 'genre' to numeric using one-hot encoding
df_filtered['price'] = pd.cut(df_filtered['price'], bins=5, labels=False)

# Remove columns with string data types
df_numeric = df_filtered.select_dtypes(exclude=['object'])

# Optimize data types
dfTemp = df_filtered.astype('float32')
df = dfTemp.dropna()
#df = df2.select_dtypes(exclude=['object'])
X = pd.get_dummies(df, columns=['price'])

# Check for non-numeric columns and convert them if necessary
# Assuming 'record_name' is the only non-numeric column you want to drop
#X = df_encoded.drop(columns=['uri','artist_id','artist_image'])

# Ensure all columns in X are numeric
# If there are other non-numeric columns, handle them here

# Train the NearestNeighbors model
knn = NearestNeighbors(n_neighbors=3, algorithm='auto').fit(X)

#add pickle file for model
with open('knn_model_spotify.pkl', 'wb') as file:
    pickle.dump(knn, file)

# Select a random product index
selected_product_index = random.randint(0, len(df) - 1)

# Get the features of the selected product
selected_product_features = X.iloc[selected_product_index].values.reshape(1, -1)

# Find the nearest neighbors
distances, indices = knn.kneighbors(selected_product_features)
recommended_product_indices = indices.flatten()
recommended_products = df.iloc[recommended_product_indices]

print("Selected Product:")
print(df.iloc[selected_product_index])
print("\nRecommended Products:")
print(recommended_products)


```
